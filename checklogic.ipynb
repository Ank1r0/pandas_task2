```python
import pandas as pd
import numpy as np
```

```python

df = pd.read_csv("data/task_2_data_ex.csv")
```

```python
print(df.head())
```

```python
def something(df):
    anchors = df[df['produced_material_release_type'] == 'FIN'].copy()

    step1 = pd.merge(
            anchors, 
            df, 
            how='left',
            left_on=['year', 'plant_id', 'component_material'], 
            right_on=['year', 'plant_id', 'produced_material'],
            suffixes=('', '_level1')
        )

        
    step2 = pd.merge(
            step1,
            df,
            how='left',
            left_on=['year', 'plant_id', 'component_material_level1'],
            right_on=['year', 'plant_id', 'produced_material'],
            suffixes=('', '_level2')
        )
        

    return step2
    
result = something(df)
result
   
```
