```python

# Table 1: Price List
prices = {
    'Item': ['Shirt', 'Shirt', 'Pants', 'Pants'],
    'Size': ['S', 'L', 'S', 'L'],
    'Price': [20, 25, 30, 35]
}

# Table 2: Current Stock
stock = {
    'Item': ['Shirt', 'Shirt', 'Pants', 'Pants'],
    'Size': ['S', 'L', 'S', 'L'],
    'In_Stock': [10, 5, 8, 2]
}

df_prices = pd.DataFrame(prices)
df_stock = pd.DataFrame(stock)

# Merging on TWO keys
res = pd.merge(df_prices, df_stock, on=['Item', 'Size'])
print(res)  
```

```python
# Left Table: Guests
df_guests = pd.DataFrame({
    'Name': ['Jai', 'Princi', 'Gaurav'],
    'Age': [27, 24, 22]
})

# Right Table: Food Preferences
df_food = pd.DataFrame({
    'Name': ['Jai', 'Princi', 'Anuj'],
    'Food': ['Pizza', 'Salad', 'Pasta']
})

# 1. INNER: Only people in BOTH lists (Jai and Princi)
inner = pd.merge(df_guests, df_food, on='Name', how='inner')

# 2. LEFT: All Guests (Gaurav stays, but has no Food -> NaN)
left = pd.merge(df_guests, df_food, on='Name', how='left')

# 3. OUTER: Everyone (Gaurav has no Food, Anuj has no Age -> both get NaNs)
outer = pd.merge(df_guests, df_food, on='Name', how='outer')

print("--- INNER ---")
print(inner)
print("\n--- LEFT ---")
print(left)
print("\n--- OUTER ---")
print(outer)
```

```python
# Players (Our "Left" Table)
df_players = pd.DataFrame({
    'Name': ['DragonSlayer', 'MagicMage', 'ShadowNinja']}, 
    index=['ID_1', 'ID_2', 'ID_3'])

# Equipment (Our "Right" Table)
# Notice: ID_2 is missing (Mage has no gear), and ID_4 is a mystery player
df_gear = pd.DataFrame({
    'Weapon': ['Steel Sword', 'Twin Daggers', 'Golden Bow']}, 
    index=['ID_1', 'ID_3', 'ID_4'])

# .join() defaults to a LEFT join
res = df_players.join(df_gear)
res
```
