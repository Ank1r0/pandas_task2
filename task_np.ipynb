 Task3: Построить прямое произведение массивов (все комбинации с каждым элементом). На вход подается двумерный массив

```python
import numpy as np
import itertools

# 1. Setup Data
data = np.arange(4) 
side = int(np.sqrt(data.size))
matrix = data.reshape(side, side)

def get_filtered_pairs(arr):
    rows, cols = arr.shape
    
    # We avoid the 'list()' function entirely here:
    coords = [p for p in itertools.product(range(rows), range(cols))]
    
    results = []
    
    # Compare every coordinate pair
    for (r1, c1), (r2, c2) in itertools.permutations(coords, 2):
        # Skip if they share a row or share a column
        if r1 == r2 or c1 == c2:
            continue
            
        results.append([arr[r1, c1], arr[r2, c2]])
        
    return results

# 2. Execute
final_result = get_filtered_pairs(matrix)

print("Matrix:")
print(matrix)
print("\nValid Pairs (Diagonal neighbors only):")
print(final_result)
```

TASK4: Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
import numpy as np

A = np.array([[7, 3, 5], [1, 1, 7], [8, 0, 8],[9,5,3]]) # Sample from your output
B = np.array([[5, 8], [9, 5]])

final_results = []

for row in A:
    matches_B0 = np.intersect1d(row, B[0])
    matches_B1 = np.intersect1d(row, B[1])
    
    if matches_B0.size > 0 and matches_B1.size > 0:
        
        all_matches = np.union1d(matches_B0, matches_B1)
        
        if all_matches.size >= 2:
            final_results.append(row)

result = np.array(final_results)
print("Strict Matching Rows:")
print(result)
```

Задача 5: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import numpy as np
from scipy.spatial.distance import cdist
import time

X = np.random.rand(500, 3)
Y = np.random.rand(400, 3)

# --- Method 1: NumPy Broadcasting ---
start_np = time.time()

# We change X from (500, 3) to (500, 1, 3)
# This forces NumPy to compare each X with every Y
dists_np = np.sqrt(np.sum((X[:, np.newaxis, :] - Y[np.newaxis, :, :])**2, axis=2))

end_np = time.time()

# --- Method 2: Scipy cdist ---
start_scipy = time.time()
dists_scipy = cdist(X, Y, metric='euclidean')
end_scipy = time.time()

print(f"NumPy time: {end_np - start_np:.5f}s")
print(f"Scipy time: {end_scipy - start_scipy:.5f}s")

print("Results match:", np.allclose(dists_np, dists_scipy))
```

```python

# A 3x3 grid representing different sections of a factory floor
# The numbers are the status codes reported by sensors
factory_floor = np.array([
    [100, 404, 100],
    [100, 100, 505],
    [404, 100, 100]
])

# Our list of "Danger" codes
critical_errors = [404, 505]

# 1. Generate the mask (The "Logic" step)
# This keeps the 3x3 shape!
error_mask = np.isin(factory_floor, critical_errors)

print("--- Error Mask (Logic) ---")
print(error_mask)
# Result: 
# [[False  True False]
#  [False False  True]
#  [ True False False]]

# 2. Use the mask to extract only the bad values
# (This will flatten the result because the output size is unknown)
found_errors = factory_floor[error_mask]
print("\n--- Specific Error Codes Found ---")
print(found_errors) # [404, 505, 404]

# 3. Use the mask to "Clean" the data
# Let's replace all errors with -1 so we can ignore them in calculations
clean_floor = np.where(error_mask, -1, factory_floor)
print("\n--- Cleaned Factory Map ---")
print(clean_floor)
```

```python
# union
arr1 = np.array([1, 2, 3])
arr2 = np.array([2, 3, 4, 5])
union = np.union1d(arr1, arr2)
print(union)  
# [1 2 3 4 5]

# intersection
intersection = np.intersect1d(arr1, arr2)
print(intersection)  
# [2 3]

# Set Difference
difference = np.setdiff1d(arr1, arr2)
print(difference)  
# [1]

# Symmetric Difference
symmetric_difference = np.setxor1d(arr1, arr2)
print(symmetric_difference)  
# [1 4 5]
```
